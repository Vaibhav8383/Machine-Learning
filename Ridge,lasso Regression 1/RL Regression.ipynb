## Monthly Fire Analysis

```python
df['Classes'] = np.where(df['Classes'].str.contains('not fire'),'not fire','fire')
```

```python
dftemp = df.loc[df['Region'] == 1]
plt.figure(figsize=(13, 6))
sns.set_style('whitegrid')
sns.countplot(x='month', hue='Classes', data=df)
plt.title('Fire Classes by Month')
plt.suptitle('Fire Analysis of Sidi- Bel region',weight= 'bold')
plt.xlabel('Months',weight='bold')
plt.ylabel('Number of fires',weight='bold')
plt.tight_layout()
plt.show()
```

```python
dftemp = df.loc[df['Region'] == 0]
plt.figure(figsize=(13, 6))
sns.set_style('whitegrid')
sns.countplot(x='month', hue='Classes', data=df)
plt.title('Fire Classes by Month')
plt.suptitle('Fire Analysis of Brejaia region',weight= 'bold')
plt.xlabel('Months',weight='bold')
plt.ylabel('Number of fires',weight='bold')
plt.tight_layout()
plt.show()
```

#### Its observed that august and september had the most number of forests fires for both regions. And from the above plot of months, we can understand few things.

#### Most of the fires happened in augustt and very high fires happened in only 3 months - June, July and August.

#### Less Fires was on September
