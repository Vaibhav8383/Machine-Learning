```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
```

```python
X,y = make_moons(n_samples=250, noise=0.05, random_state=0)
```

```python
X
```

```python
plt.scatter(X[:,0], X[:,1], c=y)
```

```python
## Feature Scaling (standardization scaling)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
DBSCAN_model = DBSCAN(eps=0.3, min_samples=5)
```

```python
DBSCAN_model.fit(X_scaled)
```

```python
DBSCAN_model.labels_
```

```python
plt.scatter(X[:,0], X[:,1], c = DBSCAN_model.labels_)
```
