### Diabetes Predictions

```python
from sklearn.datasets import load_diabetes
```

```python
dataset = load_diabetes()
```

```python
dataset
```

```python
print(dataset.DESCR)
```

```python
import pandas as pd
df_diabetes = pd.DataFrame(data=dataset.data, columns=['age', 'sex',
                 'bmi', 'bp', 's1', 's2', 's3', 's4', 's5', 's6'])
df_diabetes.head()
```

```python
X = df_diabetes
y = dataset.target
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_train.head()
```

```python
## Correlation Matrix
X_train.corr()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 8))
sns.heatmap(X_train.corr(), annot=True, cmap='coolwarm', fmt='.2f')
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor()
regressor.fit(X_train, y_train)
```

### Hyperparameter Tuning

```python
param ={
    'criterion':['squared_error', 'friedman_mse', 'absolute_error', 'poisson'],
    'splitter':['best', 'random'],
    'max_depth':[1, 2, 3, 4, 5, 10,15,20,25],
    'max_features':['auto', 'sqrt', 'log2'],
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid = GridSearchCV(regressor, param_grid=param, cv=5, scoring='neg_mean_squared_error')
```

```python
import warnings
warnings.filterwarnings('ignore')
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
## Predicitions
y_pred = grid.predict(X_test)
```

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
print(r2_score(y_test, y_pred))
print(mean_squared_error(y_test, y_pred))
print(mean_absolute_error(y_test, y_pred))
```

```python
# checking the best parameters for the Decision Tree Regressor model
selectedmodel = DecisionTreeRegressor(criterion='friedman_mse', max_depth=5, max_features='sqrt', splitter='best')
selectedmodel.fit(X_train, y_train)
```

```python
## visualizing the Decision Tree Regressor model
import matplotlib.pyplot as plt
from sklearn import tree
plt.figure(figsize=(20, 10))
tree.plot_tree(selectedmodel, filled=True)
plt.show()
```
